```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python

from statsmodels.tsa.seasonal import seasonal_decompose
file_path = "C:\\Users\\Kishan Pawar\\Downloads\\Electric_Production.csv"
df = pd.read_csv(file_path)

print(df.isnull().sum())

df['DATE'] = pd.to_datetime(df['DATE'])
df.set_index('DATE', inplace=True)

decomposition = seasonal_decompose(df['IPG2211A2N'], model='additive', period=12)


fig, (ax1, ax2, ax3, ax4) = plt.subplots(4, 1, figsize=(15, 10), sharex=True)


decomposition.observed.plot(ax=ax1, title='Original Time Series (Observed)', color='blue')
ax1.set_ylabel('Electric Production')

decomposition.trend.plot(ax=ax2, title='Trend Component', color='red')
ax2.set_ylabel('Electric Production')

decomposition.seasonal.plot(ax=ax3, title='Seasonal Component', color='green')
ax3.set_ylabel('Electric Production')


decomposition.resid.plot(ax=ax4, title='Residual Component', color='gray')
ax4.set_ylabel('Electric Production')

plt.tight_layout()
plt.show()

# Display a summary of the difference between trend and seasonality
print("\n--- Summary of Trend vs. Seasonality in the data ---")
print("Trend:")
print("The red line in the 'Trend Component' plot shows the long-term direction of the electric production data. \nThe upward slope indicates a general increase in production over time.")

print("\nSeasonality:")
print("The green line in the 'Seasonal Component' plot shows a fixed, repeating pattern. \nThe peaks and troughs indicate a predictable, recurring pattern in electric production.")

print("\nResiduals:")
print("The gray line in the 'Residual Component' plot represents the leftover noise after \nthe trend and seasonality have been removed. It shows the random fluctuations in the data that are not explained by the other two components.")
```
